```python
import numpy as np
from scipy import stats
```

# B-modes real-space

```python
SGC = 0.42
NGC = 0.11
DES = 0.87

pvals = np.array([DES, SGC, NGC])

X = -2 * np.sum(np.log(pvals))
p_combined = stats.chi2.sf(X, 2 * len(pvals))

p_combined
```

# Tangential shear (faint)

```python
SGC = 0.7
NGC = 0.11
DES = stats.chi2.sf(11, 20)

pvals = np.array([DES, SGC, NGC])

X = -2 * np.sum(np.log(pvals))
p_combined = stats.chi2.sf(X, 2 * len(pvals))

p_combined
```
